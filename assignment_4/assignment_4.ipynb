```python
import numpy as np
import math

# Parameters
P_H = 0.25
P_L = 0.75
LAMBDA_H = 10
LAMBDA_L = 7
OBS_C = 10
OBS_P = 8
N = 10**6  # Monte Carlo trials

# Counters for rejection sampling
count_obs = 0
count_high_given_obs = 0

for _ in range(N):
    # Sample steel quality
    if np.random.rand() < P_H:
        steel = 'H'
        lam = LAMBDA_H
    else:
        steel = 'L'
        lam = LAMBDA_L

    # Generate production counts
    clips = np.random.poisson(lam)
    pins = np.random.poisson(lam)

    # Keep only trials matching the observed outcome
    if clips == OBS_C and pins == OBS_P:
        count_obs += 1
        if steel == 'H':
            count_high_given_obs += 1

# Monte Carlo estimate
if count_obs > 0:
    mc_prob = count_high_given_obs / count_obs
    print(f"Trials: {N}")
    print(f"Matches of (C={OBS_C}, P={OBS_P}): {count_obs}")
    print(f"MC estimate of P(S=H | C=10, P=8): {mc_prob:.6f}")
else:
    print("No matches found — increase N.")

# Exact calculation for comparison
def poisson_pmf(k, lam):
    return (lam**k * math.exp(-lam)) / math.factorial(k)

num_h = P_H * poisson_pmf(OBS_C, LAMBDA_H) * poisson_pmf(OBS_P, LAMBDA_H)
num_l = P_L * poisson_pmf(OBS_C, LAMBDA_L) * poisson_pmf(OBS_P, LAMBDA_L)

p_exact = num_h / (num_h + num_l)

print(f"Exact value: {p_exact:.6f}")
```
